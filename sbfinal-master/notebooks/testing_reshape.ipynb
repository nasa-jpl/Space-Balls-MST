```python
import numpy as np
from SpaceBalls.utils import get_2D_to_1D_idx
from SpaceBalls.sph_meshing import get_reshaped_grid

n_lat, n_lon = 3, 5

matrix = np.random.rand(n_lat,n_lon)
reshaped_matrix_manual = np.zeros(n_lat*n_lon)

for i in range(n_lat):
    for j in range(n_lon):
        #new_idx = i*n_lon + j
        new_idx = get_2D_to_1D_idx(i,j,n_lon)
        reshaped_matrix_manual[new_idx] = matrix[i,j]

reshaped_matrix = np.reshape(matrix, n_lat*n_lon)
print(reshaped_matrix==reshaped_matrix_manual)

J_vec, I_vec, shape = get_reshaped_grid(np.arange(n_lon), np.arange(n_lat))
print(get_2D_to_1D_idx(I_vec, J_vec, n_lon)==np.arange(n_lon*n_lat))

print(np.reshape(np.reshape(matrix, n_lon*n_lat), (n_lat, n_lon))==matrix)

```
