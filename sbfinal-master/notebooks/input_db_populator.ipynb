```python
# insert some random orbits
import SpaceBalls.input_database_manager as db_manager
import numpy as np
rng = np.random.default_rng()

n = 36

for i in range(n):
    orbit_name = 'R800_' + str(i) # random orbit at 800km #i
    db_manager.insert_orbit(orbit_name,
                            {'a': 6378.14 + 800, #+ 894,      # 6878.14, (500km) #7378.14, #1000 km mean alt  #semi-maj axis in km
                             'e': 0.0,          # 0.001, # eccentricity
                             'i': 60 + 30 * rng.random(),         # 97.40,(500 km for SS orbit) #100.66,(1250 km) #99.48,(1000 km) #58.5064, #inclination
                             'o': 360 * rng.random(),         # for 21Mar to zero longitude: 358.88, #not RAAN here but long. of node: 180.0 deg = 12 pm MLTAN, 90.0 = 6am
                             'u': 0.0,          # Arg of perigee
                             'theta': 360 * rng.random(),     # true anomaly
                            }

)
```
