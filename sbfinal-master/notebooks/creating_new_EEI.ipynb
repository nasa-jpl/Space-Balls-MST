```python
# storage of r_sun in a fullly circular orbit

import sys, os
from SpaceBalls.paths import MEDIA_DIR, OUTPUT_DIR
import numpy as np
from SpaceBalls.radiation_fluxes_preprocessing import mid_day_jd_array_from_jd_interval
import SpaceBalls.radiation_settings as rad_settings
from SpaceBalls.sph_meshing import progress_bar
import config.constants
AU = config.constants.astronomical_unit()

rad_config = rad_settings.radiation_settings_from_EEI_truth_name("EEI_truth_0") # add any other here
mid_day_jd_array = mid_day_jd_array_from_jd_interval(rad_config["jd_interval"])
dir_0 = os.path.join(MEDIA_DIR, 'solar_ephemerides', rad_config["ephemerides"], 'daily_files')
os.makedirs(dir_0, exist_ok=True)
np.save(os.path.join(dir_0, 'jd_mid_day_array'), mid_day_jd_array)

rad_config_og = rad_settings.radiation_settings_from_EEI_truth_name("EEI_truth_1")
dir_0_og = os.path.join(MEDIA_DIR, 'solar_ephemerides', rad_config_og["ephemerides"], 'daily_files')

for i, jd in enumerate(mid_day_jd_array):
    r_sun_og = np.load(os.path.join(dir_0_og, 'r_Sun_ECEF_day_'+str(i)+'.npy'))
    r_sun_og_norm = np.linalg.norm(r_sun_og, axis=1)
    u_sun_og = r_sun_og / r_sun_og_norm[:, None]
    r_sun_new = u_sun_og * AU
    np.save(os.path.join(dir_0, 'r_Sun_ECEF_day_'+str(i)), r_sun_new)

    R_mat_og = np.load(os.path.join(dir_0_og, 'R_ECEF_to_SunFrame_day_'+str(i)+'.npy'))
    np.save(os.path.join(dir_0, 'R_ECEF_to_SunFrame_day_'+str(i)), R_mat_og)

    jd_array_og = np.load(os.path.join(dir_0_og, 'jd_array_day_'+str(i)+'.npy'))
    np.save(os.path.join(dir_0, 'jd_array_day_'+str(i)), jd_array_og)

    print(f"Day {i} done")
```
