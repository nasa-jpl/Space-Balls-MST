```python
import SpaceBalls.input_database_manager as input_manager
import matplotlib.pyplot as plt
import numpy as np
from SpaceBalls.plotter import Plotter
from SpaceBalls.postproc_EEI_estimation import get_radial_measurements_array, get_full_output_var_hist

orbits = ['G3', 'A1', 'F12']
SB_constellation = np.concatenate([input_manager.get_primary_keys({'orbit_name': orb,
                                                        'force_settings': 2,
                                                        'integration_settings': 0}) for orb in orbits])
acc_meas_radial = get_radial_measurements_array(SB_constellation, meas="flux")
jd_vecs =  get_full_output_var_hist(SB_constellation, 'jd_vec')

fig, ax = plt.subplots(figsize=(19,2.5))
i_max = len(jd_vecs[0])
cols = ['k', 'b', 'r']
for i in range(len(orbits)):
    ax.plot(Plotter.jd_to_datetime(jd_vecs[i][:i_max]), acc_meas_radial[i][:i_max], 
            label=orbits[i], alpha=0.2,
            lw=0.1, color=cols[i])
ax.set_ylabel('Net radial flux. (W/m$^2$)')
ax.legend(ncols=3)
ax.set_xlim([Plotter.jd_to_datetime(jd_vecs[0][0]), Plotter.jd_to_datetime(jd_vecs[0][i_max-1])])
Plotter.format_time_labels(ax)
```

```python
fig, ax = plt.subplots(figsize=(9.5,2.5))
i_start = 0
i_end = 1500
cols = ['k', 'b', 'r']
for i in range(len(orbits)):
    ax.plot(Plotter.jd_to_datetime(jd_vecs[i][i_start:i_end]), acc_meas_radial[i][i_start:i_end], 
            label=orbits[i],
            lw=1, color=cols[i])
ax.set_ylabel('Net radial flux. (W/m$^2$)')
ax.legend(ncols=3)
ax.set_xlim([Plotter.jd_to_datetime(jd_vecs[0][i_start]), Plotter.jd_to_datetime(jd_vecs[0][i_end-1])])
Plotter.format_time_labels(ax)
```
