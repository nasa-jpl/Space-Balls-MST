```python
type_names = ['thermal', 'albedo']

for j, array in enumerate([sh_arr.thermal_array, sh_arr.albedo_array]):
    for i, thermal_map in enumerate(array):

        fig, ax_coast = plt.subplots(subplot_kw={'projection': ccrs.PlateCarree()}) # PlateCarree
        ax_coast.coastlines()
        gl = ax_coast.gridlines(draw_labels=True)
        gl.top_labels = False
        gl.right_labels = False
        fig = ax_coast.get_figure()

        clm = pysh.SHCoeffs.from_array(thermal_map, normalization='unnorm')
        #clm = pysh.datasets.Earth.GGM05C()
        grid = clm.expand()

        im = ax_coast.imshow(grid.data, extent=[grid.lons().min(), grid.lons().max(), 
                                        grid.lats().min(), grid.lats().max()], 
                        cmap='coolwarm')
        fig.colorbar(im, ax=ax_coast, shrink=0.7)
        plt.title(sh_arr.all_dates[i])
        plt.savefig("sh_figures/" + type_names[j] + "_" + str(i) + ".png", dpi=300, bbox_inches='tight', 
                    facecolor='white', transparent=False)
        


    
```
