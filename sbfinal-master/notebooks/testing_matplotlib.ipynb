```python
from SpaceBalls.plotter import Plotter

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(12, 6))

# Plot your data
x = np.linspace(0, 10, 100)
ax.plot(x, np.sin(x), label='Curve A', color='blue')
ax.plot(x, np.cos(x), label='Curve B', color='red')

# ===== PRIMARY BOTTOM AXIS (Time in seconds) =====
ax.set_xlabel('Time (s)', color='black', fontsize=12)
ax.tick_params(axis='x', labelcolor='black')
ax.set_xlim([0, 10])

# ===== SECONDARY BOTTOM AXIS (Time in minutes) =====
ax_bottom2 = ax.twiny()
ax_bottom2.spines['bottom'].set_position(('outward', 40))
#ax_bottom2.spines['top'].set_visible(False)  # Hide unused spine
ax_bottom2.set_xlim(ax.get_xlim())
ax_bottom2.set_xticks(ax.get_xticks())
ax_bottom2.set_xticklabels([f'{t/60:.2f}' for t in ax.get_xticks()])
ax_bottom2.set_xlabel('Time (minutes)', fontsize=11)
ax_bottom2.tick_params(axis='x')
#ax_bottom2.spines['bottom'].set_color('green')
ax_bottom2.spines['bottom'].set_linewidth(1)
ax_bottom2.xaxis.set_ticks_position('bottom')
ax_bottom2.xaxis.set_label_position('bottom')

# ===== TOP AXIS 1 (Curve A Scale) =====
ax2 = ax.twiny()
ax2.spines['bottom'].set_visible(False)  # Hide unused spine
ax2.set_xlim(ax.get_xlim())
ax2.set_xticks(ax.get_xticks())
ax2.set_xticklabels([f'{t:.1f}' for t in ax.get_xticks()])
ax2.set_xlabel('Curve A Scale (Hz)', color='blue', fontsize=11)
ax2.tick_params(axis='x', labelcolor='blue')
ax2.spines['top'].set_color('blue')
ax2.spines['top'].set_linewidth(1)

# ===== TOP AXIS 2 (Curve B Scale) =====
ax3 = ax.twiny()
ax3.spines['top'].set_position(('outward', 40))
ax3.spines['bottom'].set_visible(False)  # Hide unused spine
ax3.set_xlim(ax.get_xlim())
ax3.set_xticks(ax.get_xticks())
ax3.set_xticklabels([f'{t*2:.1f}' for t in ax.get_xticks()])
ax3.set_xlabel('Curve B Scale (rad)', color='red', fontsize=11)
ax3.tick_params(axis='x', labelcolor='red')
ax3.spines['top'].set_color('red')
ax3.spines['top'].set_linewidth(1)

ax.legend(loc='upper right')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(10, 6))

# Plot your data
x = np.linspace(0, 10, 100)
ax.plot(x, np.sin(x), label='Curve A', color='blue')
ax.plot(x, np.cos(x), label='Curve B', color='red')

# ===== PRIMARY BOTTOM AXIS (Time in seconds) =====
ax.set_xlabel('Time (s)', color='black', fontsize=11)
ax.tick_params(axis='x', labelcolor='black')
ax.set_xlim([0, 10])

# ===== SECONDARY BOTTOM AXIS (Time in minutes) =====
ax_bottom2 = ax.secondary_xaxis('bottom', functions=(lambda x: x/60, lambda x: x*60))
ax_bottom2.spines['bottom'].set_position(('outward', 40))
#ax_bottom2.spines['top'].set_visible(False)
ax_bottom2.set_xlabel('Time (minutes)', color='green', fontsize=11)
ax_bottom2.tick_params(axis='x', colors='green')  # Use 'colors' for both ticks and labels
ax_bottom2.spines['bottom'].set_color('green')
ax_bottom2.spines['bottom'].set_linewidth(1)

# ===== TOP AXIS 1 (Curve A Scale) =====
ax.spines['top'].set_visible(False)
ax2 = ax.secondary_xaxis('top', functions=(lambda x: x * 0.1, lambda x: x / 0.1))
#ax2.spines['bottom'].set_visible(False)
ax2.set_xlabel('Curve A Scale (Hz)', color='blue', fontsize=11)
ax2.tick_params(axis='x', colors='blue')  # Use 'colors' for both ticks and labels
ax2.spines['top'].set_color('blue')
ax2.spines['top'].set_linewidth(1)

# ===== TOP AXIS 2 (Curve B Scale) =====
ax3 = ax.secondary_xaxis('top', functions=(lambda x: x * 2, lambda x: x / 2))
ax3.spines['top'].set_position(('outward', 40))
#ax3.spines['bottom'].set_visible(False)
ax3.set_xlabel('Curve B Scale (rad)', color='red', fontsize=11)
ax3.tick_params(axis='x', colors='red')  # Use 'colors' for both ticks and labels
ax3.spines['top'].set_color('red')
ax3.spines['top'].set_linewidth(1)

ax.legend(loc='upper right')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
